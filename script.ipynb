```python
from pathlib import Path

import pandas as pd


DATA_DIR = Path("data/ch2025_data_items")


def load_table(name: str, data_dir: Path = DATA_DIR) -> pd.DataFrame:
    """parquet 파일을 우선 읽고, 없으면 같은 이름의 csv를 읽습니다."""
    parquet_path = data_dir / f"{name}.parquet"
    csv_path = data_dir / f"{name}.csv"

    if parquet_path.exists():
        return pd.read_parquet(parquet_path, engine="fastparquet")
    if csv_path.exists():
        return pd.read_csv(csv_path)

    raise FileNotFoundError(f"{parquet_path} 또는 {csv_path} 파일을 찾을 수 없습니다.")


def load_all_data(data_dir: Path = DATA_DIR) -> dict[str, pd.DataFrame]:
    """폴더 안의 parquet를 우선 기준으로 모든 테이블을 로드합니다."""
    names = sorted({path.stem for path in data_dir.glob("*.parquet")})
    if not names:
        names = sorted({path.stem for path in data_dir.glob("*.csv")})

    return {name: load_table(name, data_dir) for name in names}
```

```python
dfs = load_all_data()
pd.DataFrame(
    [(name, *df.shape) for name, df in dfs.items()],
    columns=["table", "rows", "columns"],
)
```
